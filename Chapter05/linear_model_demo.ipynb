```python
import numpy as np
from scipy import stats
from matplotlib import pyplot as plt

rng = np.random.RandomState(1415)
```

```python
x = np.linspace(0.2, .8, 100)

def fn(x):
    return x

y = fn(x) + stats.norm.rvs(loc=0, scale=.05, size=x.size)
```

```python
xp = np.linspace(-10, 10, 100)
yp = fn(xp)

fig, ax = plt.subplots(figsize=(4, 4))
ax.scatter(x, y, label="Data")
ax.plot(xp, yp, c="k", label="Linear Model\n$y = mx + b$")

ax.set_xlim((0, 1))
ax.set_ylim((0, 1))

ax.set_xlabel("Feature [x]")
ax.set_ylabel("Target [y]")
ax.legend(loc="upper left")
fig.tight_layout()
fig.savefig("linearmodel.pdf")
fig.show()
```
